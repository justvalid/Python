```python
# Grading system using conditional statements

score = int(input("Enter score to determine your grade: "))

# conditions for the grading
if score >= 70 and score <= 100:
    print("Grade A")

elif score >= 60 and score < 70:
    print("Grade B")

elif score >=50 and score < 60:
    print("Grade C")

elif score >=45 and score < 50:
    print("Grade D")

elif score >=40 and score < 45:
    print("Grade E")

elif score <=39 and score < 40:
    print("Grade F")

else:
    print("Invalid score")
```

```python
"""Data structures in python"""

# Defining a List in python
names = ["Glory", "Jackson", "Charles"]
print(names)

# Accessing a list
print(names[1])
print(names[2])

# Modification
names[1] = "Daniel"
print(names)

# Removing an item from a list
names.remove("Glory")
print(names)

# Adding to a list
names.append("Glory")
print(names)
```

```python
# Tuples
readings = (2, 5, 3.3, 2.1)
print(readings)

print(readings[3])
```

```python
# Sets

numbers = {2, 4, 8, 10, 5, 16, 2, 2, 2}
print(numbers)


# Union
set_a = {2, 4, 5}
set_b = {2, 3, 5}

print(set_a | set_b)

# Intersection
print(set_a & set_b)

# difference
print(set_a - set_b)
```

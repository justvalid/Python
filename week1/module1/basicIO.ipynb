```python
"""
Topic: Basic Input and Output in Python
"""

# taking name input from a user
name = input("Enter your name:")

# taking age input from a user
age = int(input("Enter your age:"))

gender = input("Gender: ")
dept = input("Enter your Department: ")

print(name)
print(age)
print(gender)
print(dept)
```

```python
"""STRING CONCATENATION"""

"""METHODS 1 USING THE '+' OPERATOR"""
country = input("Enter your country: ")
print("I am a citizen of: " + country)

"""METHODS 2 USING THE ',' OPERATOR"""
country = input("Enter your country: ")
print("I am a citizen of:", country)

"""METHODS 3 USING F-STRINGS"""
country = input("Enter your country: ")
print(f"I am a citizen of: {country}")
```

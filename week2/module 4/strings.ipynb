```python
"""
Python Programming Practice: Simple Email Spam Filter


 *Objective* 
Write a simple Python script that scans an email body for a specific spam keyword and prints whether the email is spam or not.

 *Strict Rules & Constraints* 

To help you focus purely on string manipulation and basic data types, you must follow these strict rules:
• 🛑 No functions (Do not use def).
• 🛑 No control flow statements (Do not use if, else, for, or while).
• 🟢 Use the membership operator (in) to perform the check directly.

 *Instructions* 

1. Get Input: Ask the user to enter or paste the content of an email.
2. Set a Keyword: Define a single spam word to look for (e.g., "free").
3. Handle Capitalization: Convert the user's input to lowercase so the check works even if they type "FREE" or "Free".
4. Check & Print: Check if your keyword is inside the text and print the direct True or False result.

"""

# Get Input: Ask the user to enter or paste the content of an email.

email_body = input("Enter email message body: ")

# Define keywords to look for

spam_words = "Free"

# Convert the user's input to lowercase
spam_check = email_body.lower()

# print the direct True or False result for each spam word
is_spam = spam_words in spam_check
# is_spam = spam_words in spam_check

print(is_spam)
```
